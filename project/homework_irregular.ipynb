# Introduction

In this project you will study one year of sea level, measured a few times per day, at irregular times, in the harbour of Boston, on the east coast of the United States. The sea there goes up and down by about three metres twice a day, and a few times a year a storm pushes it high enough to flood the streets near the harbour. The measured level is the sum of several effects that nobody can observe separately. Your job is to separate them, with a model that also says how sure it is about each one.

Before looking at any data, you need to understand what makes the sea go up and down.

## What moves the sea

**1. The tide.** The Moon pulls on everything on Earth, and its pull gets weaker with distance. The ocean on the side facing the Moon is closer to it than the rest of the Earth, so it is pulled a bit more and rises towards the Moon. The ocean on the opposite side is further away, so it is pulled a bit less than the Earth. The Earth moves slightly away from it, and this water rises too. The result is two zones of high water, one facing the Moon and one on the opposite side, with lower water in between.

The Earth spins on itself about once a day, and every harbour spins with it. The two high zones do not spin with the Earth. They stay aligned with the Moon, because the Moon is what creates them. So as the Earth spins, a harbour is carried from a high zone to a low zone, then to the second high zone, then to a low zone again, and then it is back facing the Moon. This takes a bit more than a day, because while the Earth spins, the Moon also moves along its orbit in the same direction. It goes around the Earth in 27.3 days, so it moves by about 13 degrees per day, and the Earth needs about 50 extra minutes to catch up with it. The Moon therefore passes over the same harbour every 24 h 50 min, and the harbour gets a high water every 12 h 25 min. Your measurements are too far apart to see this directly, but the periodogram of Question 2 will find it. This main part of the tide is called **M2**.

The Sun does exactly the same, with about half the strength of the Moon. The Sun comes back over the same place every 24 h, which is the definition of a day, so its tide has a period of exactly 12 h. It is called **S2**. Because 12 h and 12 h 25 min are close but not equal, the two tides slowly go in and out of step. When they are in step they add up and the tide is large. When they are out of step they partly cancel and the tide is small. They come back in step every 14.77 days, which is why the size of the tide grows and shrinks over about two weeks. The animation below shows this.

<img src="tides.gif" width="800">

Other effects complete this picture. The distance between the Earth and the Moon changes over 27.55 days, so the pull of the Moon is not always the same, and one small tide out of two is smaller than the other. The orbit of the Moon is also tilted. The Moon is usually not above the equator but north or south of it, up to about 28 degrees. The two high zones are then centred north and south of the equator, so a harbour passes close to the centre of one zone and far from the centre of the other, and its two high waters of the day are unequal. This difference disappears every two weeks, when the Moon crosses the equator.

<img src="tilt_explained.png" width="900">

This has a consequence that surprises everyone the first time. A periodogram does not count high waters. It rebuilds the curve as a sum of sine waves and tells you how big each one must be. A day with one big and one small high water cannot be rebuilt from waves with two cycles per day only, because they always give two equal highs. It needs a wave with one cycle per day as well, the dotted curve in the figure. So your periodogram will show lines near one cycle per day, even though the sea clearly rises twice a day.

Each of these effects is written as one more sine wave with its own period, and each of these sine waves is called a **constituent**. M2 is the largest almost everywhere, but which one comes next depends on the harbour. In Boston it is the one caused by the distance of the Moon, not S2. In Question 4 you will compute the periods of about ten constituents from four astronomical numbers.

This leads to the most important fact of the project. The **periods** of the constituents come from astronomy and are known very precisely, before looking at any data. Their **size and timing** in a given harbour are unknown. They depend on the shape of the coast and the depth of the sea, so the same Moon gives a tide of 0.3 m in the Mediterranean, 3 m in Boston, and 16 m in the Bay of Fundy, a few hundred kilometres northeast of Boston. Size and timing must be learned from the data.

One more effect belongs to the tide, and it matters in Boston. Far out at sea, the tide is close to a pure sine wave. In a shallow harbour, a wave travels faster where the water is deeper, so the top of the wave travels faster than the bottom and the shape is slightly distorted. In Boston you will not see this distortion by eye, but its first consequence, a constituent at twice the frequency of M2 called M4, is in the table of Question 4. The wave still repeats every 12 h 25 min. And a signal that repeats every $T$ hours without being a pure sine is a sum of sines with periods $T$, $T/2$, $T/3$, and so on. This is a Fourier series. So the harbour itself creates extra constituents, at 6 h 13 min and 4 h 8 min, even though nothing in the sky moves at those speeds. They belong to the tide because they repeat with it, and their periods are known exactly. Not every line in your periodogram comes from the sky.

**2. The storm surge.** The weather also moves the sea, in two ways. The air has a weight and presses on the water. When the air pressure drops, the sea rises by about 1 cm for each hPa. A storm at 980 hPa, compared with a normal 1013 hPa, therefore lifts the sea by about 33 cm, and you will be able to check this number with the data. Wind blowing towards the coast also pushes water against it. This rise of the sea caused by the weather is called the **surge**. It does not repeat, so no calendar can predict it. It builds up over a few hours when a storm arrives, and goes back to zero over a day or two once the storm has passed. The surge is what floods harbours, and it is the reason models like yours exist.

**3. The yearly cycle.** Water expands when it gets warmer, and winds and sea currents change with the seasons. In Boston this makes the average sea level go up and down by about 10 cm over the year, higher from May to October and lower in winter. This has nothing to do with the Moon, but it also repeats with a known period, one year. So it is handled exactly like a tidal constituent.

**4. The fast movements.** Many things move the sea within an hour, such as waves, or the water sloshing back and forth inside the harbour. With measurements several hours apart, these look like random noise from one measurement to the next. The error of the instrument itself adds a little more. The noise term of your model collects both.

## Why this is a data science problem

The gauge only measures the sum of these four effects. No instrument measures the tide or the surge separately, so for this split there is no ground truth, not here and not anywhere. Even the tide prediction published by the agency running the gauge, which you also get, is not a measurement but the output of another model.

This is exactly the situation of the latent variable models of the course. The tide, the surge and the yearly level are latent states that change over time. The measurement is their sum plus noise. What you want is the posterior distribution of these states given the data.

Separating them matters in practice. Tide tables can be printed years in advance, because the tide repeats. Flood warnings are about the surge, which does not. And the highest water levels do not simply come from the largest storms. They come from a storm that arrives on a day of large tide.

The usual method, used for more than a century, fits a sum of sine waves at the known periods by least squares, subtracts it, and calls what is left the surge. It works, but it gives no uncertainty on the surge at a given hour, it assumes that the size of each constituent never changes, and it cannot forecast the surge of the next two days with error bars. A state-space model does all of this, and this is what you will build.

You will first explore the data, then build the model from the physics above, compute the posterior of the hidden components with the Kalman filter and smoother, estimate the parameters by maximum likelihood, and finally separate the tide, the surge and the mean level.

# Build: a state-space model of the water level

## Continuous time

We first describe the system in continuous time, and discretize it afterwards, as recommended in [Lecture 5](https://glouppe.github.io/dats0001-foundations-of-data-science/?p=lecture5.md#48).

The **tide** is written as in the introduction, as a sum of sine waves, one per constituent. With $J$ constituents,
$$\text{tide}(t) = \sum_{j=1}^{J} \alpha_j \cos(\omega_j t - \phi_j),$$
where constituent $j$ has a known angular frequency $\omega_j$, which you will compute in Question 4, an unknown amplitude $\alpha_j$, its size, and an unknown phase $\phi_j$, its timing. Which constituents to include, and therefore the value of $J$, is your choice in Question 5.

To use the tide in a state-space model, each constituent must be written as in Lecture 5, as a state that moves from one step to the next by a linear rule, plus a little noise. A cosine is not directly of that form, but it becomes so with a simple picture.

Imagine a point that turns around a circle of radius $\alpha_j$ at constant speed, one full turn per period of the constituent. At time $t$ its angle is $\omega_j t - \phi_j$. Its horizontal coordinate is
$$c_j(t) = \alpha_j \cos(\omega_j t - \phi_j),$$
which is exactly the contribution of constituent $j$ to the water level. Its vertical coordinate is
$$e_j(t) = \alpha_j \sin(\omega_j t - \phi_j).$$

<img src="rotating_point.png" width="900">

The water level only uses $c_j$. We still need $e_j$, because $c_j$ alone does not say where the point goes next. The same horizontal position is reached once while $c_j$ goes up and once while it goes down, and only $e_j$ tells which.

The pair $r_j(t) = (c_j(t), e_j(t))$ is the position of the point, and it contains the size and the timing of the constituent. The size $\alpha_j$ is the distance of the point to the centre, and the timing $\phi_j$ fixes where the point is on the circle at a given time. So $\alpha_j$ and $\phi_j$ do not need to be parameters of the model. We put $r_j$ in the state, the Kalman filter estimates the position of the point, and the size and the timing come with it.

Time is measured in hours, so $\omega_j = 2\pi / (\text{period in hours})$, in radians per hour. For M2, $\omega_j = 2\pi / 12.42$. Taking the derivative of $c_j$ and $e_j$ with respect to time gives $\frac{dc_j}{dt} = -\omega_j e_j$ and $\frac{de_j}{dt} = \omega_j c_j$, a linear rule. Adding small random pushes, the equation of the turning point in continuous time is
$$dr_j(t) = \Omega_j \, r_j(t) \, dt + \sigma_\text{tide} \, dW(t), \qquad \Omega_j = \begin{bmatrix} 0 & -\omega_j \\ \omega_j & 0 \end{bmatrix}.$$
The first term turns the point at speed $\omega_j$. The second term gives it small random pushes, which slowly change its distance to the centre and its position on the circle, so the size and the timing of the constituent can drift a little during the year.

With $\sigma_\text{tide} = 0$ there are no pushes, and the point turns forever on the same circle. This is a sine wave of fixed size and timing, as in the classical method of the introduction.

The **surge** $s(t)$, the rise of the sea caused by the weather, is pushed away from zero when a storm arrives and relaxes back once it has passed. This is the same Ornstein-Uhlenbeck process that described the wolf in [Lecture 5](https://glouppe.github.io/dats0001-foundations-of-data-science/?p=lecture5.md#45), in one dimension and with equilibrium point $0$,
$$ds(t) = -\kappa \, s(t) \, dt + \sigma_s \, dW(t),$$
where $1/\kappa$ [h] is the relaxation time of the surge, the time it needs to shrink by a factor $e$ once the weather is calm, and $\sigma_s^2 / (2\kappa)$ is its variance in the long run.

In the model, the effect of the air pressure is written separately, as $\beta \tilde{p}_t$ in the observation below. So $s(t)$ is the part of the surge that the local air pressure does not explain, for example the push of the wind, and the total surge is $s(t) + \beta \tilde{p}(t)$.

The **yearly cycle** is the slow rise and fall of the average sea level over the year, described in the introduction. It repeats every year, so it can be written exactly like the tide, as sine waves with known periods. One sine wave with a period of one year gives a very regular shape, six months up and six months down. The real yearly cycle does not have to be that regular, so we add a second sine wave with a period of half a year. Together, the two can make a shape that rises faster than it falls, or stays high for longer than it stays low. This is the same idea as the shallow water effect of the introduction, with periods $T$ and $T/2$. These two sine waves are simply two more constituents in the sum above, each with its own turning point. They follow the same equation as the other constituents, $dr_j(t) = \Omega_j \, r_j(t) \, dt + \sigma_\text{tide} \, dW(t)$, with $\omega_j = 2\pi / (1 \text{ year})$ for the first and $\omega_j = 2\pi / (\text{half a year})$ for the second. Only the speed of the turning point changes. In Question 4 they are called Sa and Ssa.

The **mean level** is a constant $\ell$, the shift between the zero level of the station and the average level of the sea in 2024. Since it does not change, its equation is simply $d\ell(t) = 0$. A mean level that drifts slowly over time would be a natural alternative, but it is not used here.

Each $dW(t)$ above is its own Brownian motion, independent of all the others, and the two coordinates of each $r_j$ receive independent pushes.

## Discrete time

The data arrive at irregular times. In the continuous equations above, $t$ was the time in hours. From now on, $t$ counts the measurements, and $\Delta t$ is the time between measurement $t-1$ and measurement $t$, between 3 and 9 h. We now write how each component moves from one measurement to the next. $\Delta t$ changes from one step to the next, so the transitions below change too, as for the wolf in [Lecture 5](https://glouppe.github.io/dats0001-foundations-of-data-science/?p=lecture5.md#47). Each equation above can be solved exactly over one step, with no approximation. For the surge, this is the exact discretization of the Ornstein-Uhlenbeck process in [Lecture 5](https://glouppe.github.io/dats0001-foundations-of-data-science/?p=lecture5.md#47), with $\mu = 0$. For the tide, the solution is a rotation. You do not need to derive it, it is given below.

- **Tide.** In one step, each turning point turns by the angle $\omega_j \Delta t$, and the random pushes add a small noise,
$$r_{j,t} = \text{Rot}(\omega_j \Delta t)\, r_{j,t-1} + w_{j,t}, \qquad w_{j,t} \sim \mathcal{N}\left(0, \sigma_\text{tide}^2 \Delta t \, I_2\right), \qquad \text{Rot}(a) = \begin{bmatrix} \cos a & -\sin a \\ \sin a & \cos a \end{bmatrix}.$$
This holds for all $J$ constituents, including the two of the yearly cycle.
- **Surge.** In one step, the surge keeps only a fraction $e^{-\kappa \Delta t}$ of its value and receives a new random push,
$$s_t = e^{-\kappa \Delta t}\, s_{t-1} + \eta_t, \qquad \eta_t \sim \mathcal{N}\left(0, \frac{\sigma_s^2}{2\kappa}\left(1 - e^{-2\kappa \Delta t}\right)\right).$$
- **Mean level.** It does not move, $\ell_t = \ell_{t-1}$.

Stacking all the components into one vector, the latent state
$$z_t = (c_{1,t}, e_{1,t}, \ldots, c_{J,t}, e_{J,t}, s_t, \ell_t) \in \mathbb{R}^{2J + 2},$$
the three transitions above (tide, surge, mean level) become a single linear Gaussian transition, as in [Lecture 5](https://glouppe.github.io/dats0001-foundations-of-data-science/?p=lecture5.md#20) with $b = 0$,
$$p(z_t | z_{t-1}) = \mathcal{N}(z_t | A z_{t-1}, Q),$$
where $A$ and $Q$ are computed with the $\Delta t$ of that step, $A$ contains the rotations, the factor $e^{-\kappa \Delta t}$ and a $1$ for the mean level, and $Q$ contains the noise variances.

The measured water level $x_t$ is the sum of the components, plus the effect of the air pressure and a noise term that collects the fast movements and the error of the instrument,
$$x_t \mid z_t \sim \mathcal{N}(H z_t + \beta \tilde{p}_t, R), \qquad R = \sigma_x^2.$$
Here $H$ is a single row, since $x_t$ is a single number, with one entry per component of $z_t$. It has a $1$ in front of each $c_{j,t}$, of $s_t$ and of $\ell_t$, and a $0$ in front of each $e_{j,t}$, which is not part of the water level, so that
$$H z_t = \sum_{j=1}^{J} c_{j,t} + s_t + \ell_t.$$
The air pressure minus its average over the year is $\tilde{p}_t = p_t - \bar{p}$ [hPa], a known input called a covariate, and $\beta$ [m/hPa] is an unknown parameter.

The state contains exactly what is hidden and needed, either to build the water level now or to move it forward to the next measurement. Each $c_{j,t}$ is part of the water level. Each $e_{j,t}$ is not, but without it $c_{j,t}$ cannot be moved forward, since we would not know whether it goes up or down. The surge $s_t$ is hidden and has memory, since a surge at one moment is still partly there a few hours later. The mean level $\ell_t$ is unknown, and putting it in the state lets the Kalman filter estimate it. The air pressure is not in the state, because it is measured, and neither is the observation noise, because it has no memory from one measurement to the next.

The parameters of the model are
$$\theta = (\kappa, \sigma_s, \sigma_\text{tide}, \sigma_x, \beta).$$
